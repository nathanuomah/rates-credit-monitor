```python
def bond_price(face, coupon_rate, ytm, years, freq=1):
  coupon = face * coupon_rate / freq
  rate = ytm / freq
  periods = years * freq
  return sum([coupon / (1 + rate) ** t for t in range(1, periods + 1)]) + face / (1 + rate) ** periods
print(bond_price(100, 0.03125, 0.03139, 10))           # no freq, so the default of 1 is used
print(bond_price(100, 0.03125, 0.03139, 10, freq=2))   # semi-annual

def cash_flows(face, coupon_rate, years, freq=1):
  coupon = face * coupon_rate / freq
  periods = years * freq
  flows = [coupon for t in range(1, periods + 1)]
  flows[-1] += face
  return flows

print(cash_flows(100, 0.03125, 3))
print(cash_flows(100, 0.03125, 3, freq=2))

def bond_price(face, coupon_rate, ytm, years, freq=1):
    flows = cash_flows(face, coupon_rate, years, freq)
    rate = ytm / freq
    return sum([cf / (1 + rate) ** t for t, cf in enumerate(flows, start=1)])

print(bond_price: "bond price"(100, 0.03125, 0.03139, 10))
print(bond_price: "bond price" (100, 0.03125, 0.03139, 10, freq=2))
```
